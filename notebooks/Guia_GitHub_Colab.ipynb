# Mini guía · Tu primera experiencia en GitHub desde Colab
**Curso de Capacitación en Infraestructura Computacional para Finanzas Cuantitativas · Tema 2: Ambientes de desarrollo y gestión de entornos**

**Qué vas a hacer:** clonar el repositorio del curso, crear tu propia rama, guardar tu `requirements.txt` en un commit, enviarlo a GitHub con `push` y abrir un **pull request**.

**Antes de empezar**
1. Entra a GitHub con tu **cuenta educativa** y abre el repositorio del curso (tu docente comparte el enlace).
2. Crea un **token de acceso personal**: GitHub → *Settings → Developer settings → Personal access tokens*. Dale permiso de lectura y escritura sobre el repositorio. Si tu organización usa inicio de sesión único (SSO), pulsa **Configure SSO → Authorize** junto a tu token.
3. En Colab, abre el panel **Secretos** (ícono de llave, barra izquierda), agrega un secreto llamado `GITHUB_TOKEN` con el valor del token y **activa el acceso para este cuaderno**.

> **Importante:** el token es una contraseña. No lo pegues en el cuaderno, no lo compartas y no ejecutes `!git remote -v` (mostraría el token en pantalla).

## Paso 0 · Tus datos
Completa los valores. El apellido va en minúsculas, sin tildes ni espacios.

```python
ORG = "ORGANIZACION"            # organización o usuario dueño del repositorio (te lo indica el docente)
REPO = "REPOSITORIO"            # nombre del repositorio del curso
NOMBRE = "Tu Nombre"
CORREO = "tu_correo@ejemplo.edu"
APELLIDO = "apellido"
```

## Paso 1 · Configurar Git y clonar (`git clone`)
`git config` le dice a Git quién eres (aparecerá en tus commits). `git clone` descarga una copia del repositorio a esta sesión de Colab. El símbolo `!` ejecuta un comando de terminal y `%cd` entra a la carpeta del repositorio.

```python
from google.colab import userdata
token = userdata.get("GITHUB_TOKEN")

!git config --global user.name "{NOMBRE}"
!git config --global user.email "{CORREO}"
!git clone https://{token}@github.com/{ORG}/{REPO}.git
%cd {REPO}
```

## Paso 2 · Mirar dónde estás (`git status` y `git log`)
**Qué observar:** en qué rama estás (normalmente `main`) y los últimos commits del repositorio.

```python
!git status
!git log --oneline -n 5
```

## Paso 3 · Crear tu rama (`git switch -c`)
Una **rama** es una copia paralela para trabajar sin tocar la versión principal.

```python
!git switch -c {APELLIDO}-requirements
```

## Paso 4 · Crear tu archivo
Generamos tu `requirements.txt` (la lista de librerías con su versión exacta) dentro de la carpeta `entornos/` del repositorio. Cada estudiante usa un archivo con su apellido, para no pisarse entre compañeros.

```python
import importlib.metadata as md
from pathlib import Path

libs = ["numpy", "pandas", "scikit-learn"]
lineas = [f"{l}=={md.version(l)}" for l in libs]
Path("entornos").mkdir(exist_ok=True)
archivo = Path(f"entornos/requirements_{APELLIDO}.txt")
archivo.write_text("\n".join(lineas) + "\n")
print(archivo.read_text())
```

## Paso 5 · Tu primer commit (`git add` y `git commit`)
- `git add` elige qué cambios irán en el commit.
- `git commit -m "mensaje"` guarda una versión con un mensaje que explica el cambio.

**Qué observar:** `git status` pasa de "sin seguimiento" a "listo para commit", y el commit aparece en el historial.

```python
!git status
!git add entornos/
!git commit -m "Agrega requirements de {APELLIDO}"
!git log --oneline -n 3
```

## Paso 6 · Enviar a GitHub (`git push`)
`git push` sube tus commits a GitHub. La opción `-u` deja la rama conectada para los próximos envíos.

```python
!git push -u origin {APELLIDO}-requirements
```

## Paso 7 · Abrir un pull request (en la página de GitHub)
Un **pull request** (PR) es una solicitud para que revisen tus cambios y los incorporen a la rama principal. **No es un comando**: se hace en la web.

1. Entra al repositorio en GitHub. Verás un aviso amarillo con tu rama y el botón **Compare & pull request**.
2. Escribe un título (por ejemplo, *Requirements de Apellido*) y una descripción breve.
3. Pulsa **Create pull request**.
4. Tu docente revisa los cambios y, si todo está bien, hace el **merge** (incorpora tu rama a `main`).

## Paso 8 · Traer lo aprobado (`git switch` y `git pull`)
Cuando tu PR ya fue incorporado, vuelve a `main` y trae los cambios nuevos de GitHub.

```python
!git switch main
!git pull
!ls entornos
```

## Atajo con la interfaz de Colab: guardar un cuaderno en GitHub
Para guardar **este cuaderno** (no cualquier archivo) sin escribir comandos:
1. En Colab: **Archivo → Abrir cuaderno → pestaña GitHub**. Marca *Incluir repositorios privados* y autoriza a Colab en la ventana de GitHub. Si tu organización lo pide, un administrador debe aprobar la aplicación de Colab.
2. **Archivo → Guardar una copia en GitHub**: elige el repositorio, la rama, la ruta del archivo y escribe el mensaje del commit.

Cada vez que guardas así, Colab crea un **commit nuevo** en el repositorio: no hay guardado automático hacia GitHub.

## Si algo falla
| Mensaje | Qué significa | Qué hacer |
|---|---|---|
| `Authentication failed` o `403` | El token no tiene permiso o no está autorizado para SSO | Revisa los permisos del token y pulsa *Configure SSO → Authorize* |
| `Repository not found` | URL mal escrita o no tienes acceso | Revisa `ORG` y `REPO`, y que aceptaste la invitación |
| `nothing to commit` | No hay cambios para guardar | Verifica que el archivo esté dentro de la carpeta del repositorio |
| `rejected` (non-fast-forward) | Hay cambios nuevos en GitHub | Ejecuta `!git pull` y vuelve a hacer `push` |
| `Please tell me who you are` | Falta configurar nombre y correo | Ejecuta de nuevo el Paso 1 |
